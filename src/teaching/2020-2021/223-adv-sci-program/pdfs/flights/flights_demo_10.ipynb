```python
%matplotlib inline

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
flights = pd.read_csv('flight_data/nyc_flights.csv')
flights.head()
```

```python
fig1 = plt.figure()

plt.hist(flights['air_time'], bins=50, density=True, label="Histogram", color='red', alpha=0.3)

xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)

mean = flights['air_time'].mean()
std = flights['air_time'].std()

pdf = norm.pdf(x, mean, std)
plt.plot(x, pdf, label="Dist. Line")

plt.legend()
```

```python
top_5_dest = flights['dest'].value_counts().head()
print(top_5_dest)
```

```python
fig2 = plt.figure()

plt.bar(top_5_dest.index, top_5_dest.values, color=['red', 'yellow', 'blue', 'pink', 'black'])
plt.locator_params(axis='y', nbins=10)
plt.title('Top 5 Destinations from NYC')
plt.xlabel("Airport Codes")
plt.ylabel("# of Flights")
```
