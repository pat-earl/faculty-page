# Pandas HW3 Example

This is an example of completing the requirements for homework 3 with just using pandas. 

This will import the *netflix_titles.csv* as a Pandas **DataFrame**.

```python
import pandas as pd
import numpy as np
```

```python
# This will import the csv file into a dataframe called netflix
netflix = pd.read_csv('netflix_titles.csv', index_col=0)
```

```python
netflix.head(10)
```

```python
tv_show = netflix[netflix['type'] == "TV Show"]
# del tv_show['type']
# del tv_show['release_year']
tv_show = tv_show[['title', 'director', 'country', 'date_added', 'duration']]
tv_show['duration'] = tv_show['duration'].str.split(' ').str[0]
tv_show['duration'] = tv_show['duration'].astype('int64')
```

```python
tv_show.dtypes
```

```python
tv_show.sort_values('duration', ascending=False, inplace=True)
```

```python
tv_show.to_json('netflix_shows2.json', indent=4, orient='records')
```
