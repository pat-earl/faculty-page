```python
%matplotlib inline

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
flights = pd.read_csv('flight_data/nyc_flights.csv')

#flights.describe()
flights.head()
# flights['air_time'].describe()
```

```python
fig1 = plt.figure()

plt.hist(flights['air_time'], density=True, bins=50, label='Histogram', alpha=0.3, color='red')

xmin, xmax = plt.xlim()

x = np.linspace(xmin, xmax, 100)

mean = flights['air_time'].mean()
std = flights['air_time'].std()

pdf = norm.pdf(x, mean, std)
plt.plot(x, pdf, label='Norm. Dist.', color='black')


plt.legend()
```

```python
top_5_dests = flights['dest'].value_counts().head()
print(top_5_dests)
```

```python
fig2 = plt.figure()

plt.bar(top_5_dests.index, top_5_dests.values)


plt.xlabel('Dest. Airport Codes')
plt.ylabel('# of Flights')
plt.title('Top 5 Dests. from NYC Airports')
```
