This is an example showing how to complete homework 3 using just pandas. 

1. Import *netflix_titles.csv* file into a dataframe.
    * Make *show_id* the index using `index_col`. 
2. Create a new dataframe of *just* TV shows.
3. Remove the extra columns to keep only *title, director, country, date_added, and duration*.
4. Convert duration to a int for sorting
5. Sort the tv_shows dataframe using `sort_values()`
6. Export it to json

```python
import pandas as pd
import numpy as np
```

```python
# This will open the the netflix_titles.csv file and make it a dataframe
netflix = pd.read_csv('netflix_titles.csv', index_col='show_id')
```

```python
netflix.head()
```

```python
tv_shows = netflix[netflix['type'] == 'TV Show']
tv_shows = tv_shows[['title', 'director', 'country', 'date_added', 'duration']]
```

```python
tv_shows.head()
```

```python
tv_shows['duration'] = tv_shows['duration'].str.split(' ').str[0]
tv_shows['duration'] = tv_shows['duration'].astype('int64')
```

```python
tv_shows.dtypes
```

```python
tv_show = tv_shows.sort_values('duration', ascending=False)
```

```python
tv_show.to_json('netflix_shows.json', indent=4, orient='records')
```
